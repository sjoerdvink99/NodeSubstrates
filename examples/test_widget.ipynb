# Test Widget Rendering

Simple test to check if anywidget renders at all.

```python
import anywidget
import traitlets

class SimpleWidget(anywidget.AnyWidget):
    _esm = '''
    function render({ model, el }) {
        el.innerHTML = '<div style="padding: 20px; background: #e0f0ff; border: 2px solid blue; border-radius: 8px;"><h2>✓ anywidget works!</h2><p>If you see this, the widget rendering is functional.</p></div>';
    }
    export default { render };
    '''

SimpleWidget()
```

```python
# Test D3 import
class D3TestWidget(anywidget.AnyWidget):
    _esm = '''
    import * as d3 from "https://esm.sh/d3@7";
    
    function render({ model, el }) {
        console.log("D3 loaded:", d3);
        const svg = d3.select(el)
            .append("svg")
            .attr("width", 200)
            .attr("height", 100);
        
        svg.append("rect")
            .attr("width", 200)
            .attr("height", 100)
            .attr("fill", "#4CAF50");
        
        svg.append("text")
            .attr("x", 100)
            .attr("y", 55)
            .attr("text-anchor", "middle")
            .attr("fill", "white")
            .attr("font-size", "16px")
            .text("✓ D3 works!");
    }
    export default { render };
    '''

D3TestWidget()
```
